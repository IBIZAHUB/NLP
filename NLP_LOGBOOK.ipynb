```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')

text = "This is a simple example showing stopwords removal in Python."

words = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word.lower() not in stop_words]

print("Original Words:")
print(words)

print("\nAfter Stopwords Removal:")
print(filtered_words)
```

```python
import nltk
from nltk.util import bigrams, trigrams
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')

text = "Natural language processing is interesting"
tokens = word_tokenize(text)

print("Bigrams:")
for bg in bigrams(tokens):
    print(bg)

print("\nTrigrams:")
for tg in trigrams(tokens):
    print(tg)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

text = "Natural Language Processing is a fascinating field of Artificial Intelligence."

# Tokenization
tokens = word_tokenize(text)

# POS Tagging
pos_tags = pos_tag(tokens)

# Analysis
word_count = len(tokens)
char_count = len(text.replace(" ", ""))

print("Sentence:", text)
print("Word Count:", word_count)
print("Character Count:", char_count)
print("\nTokens:")
print(tokens)

print("\nPart-of-Speech Tags:")
for word, tag in pos_tags:
    print(f"{word}: {tag}")
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')

sentence = "The quick brown fox jumps over the lazy dog"

tokens = word_tokenize(sentence)

tagged_sequence = pos_tag(tokens)

print("Tokens:", tokens)
print("\nSequence Labeling Output (Word → POS Tag):")
for word, tag in tagged_sequence:
    print(f"{word} → {tag}")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Barack Obama was born in Hawaii and worked at the United States government in Washington DC in 2009."

doc = nlp(text)

print("Named Entities Found:\n")
for ent in doc.ents:
    print(ent.text, "→", ent.label_)
```

```python
import spacy

# Load NLP model
nlp = spacy.load("en_core_web_sm")

# Sentences
texts = [
    "The administrator updated student records",
    "Machine learning improves language processing"
]

# Process each sentence
for text in texts:
    doc = nlp(text)
    print("\nSentence:", text)
    print("Dependency Structure:")

    for token in doc:
        print(token.text, "→", token.dep_, "→", token.head.text)
```

```python
!python -m spacy download en_core_web_md
import spacy

# Load model with word vectors
nlp = spacy.load("en_core_web_md")

# Sentence pairs
pair_1 = ("Machine learning improves language processing",
          "AI techniques enhance natural language understanding")

pair_2 = ("The cat is sleeping on the chair",
          "Economic policies affect global trade markets")

# Process sentences
doc1 = nlp(pair_1[0])
doc2 = nlp(pair_1[1])

doc3 = nlp(pair_2[0])
doc4 = nlp(pair_2[1])

# Similarity scores
sim1 = doc1.similarity(doc2)
sim2 = doc3.similarity(doc4)

print("Pair 1 Similarity (Related Sentences):", sim1)
print("Pair 2 Similarity (Unrelated Sentences):", sim2)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag

text = "Natural Language Processing helps computers understand language."

# Tokenization
words = word_tokenize(text)
print("Tokens:", words)

# Stopword Removal
stop_words = set(stopwords.words('english'))
filtered = [w for w in words if w.lower() not in stop_words]
print("Filtered:", filtered)

# POS Tagging
pos = pos_tag(filtered)
print("POS Tags:", pos)
```

```python
print("Welcome to Student Help Chatbot")

while True:
    user = input("You: ")

    if "hello" in user.lower():
        print("Bot: Hello Student")

    elif "exam" in user.lower():
        print("Bot: CAT 1 starts next week")

    elif "bye" in user.lower():
        print("Bot: Goodbye")
        break

    else:
        print("Bot: I do not understand")
```

```python

print("=== Student Academic Assistant Chatbot ===")
print("Type 'exit' to end the chat\n")

while True:
    user_input = input("You: ").lower()

    # Exit condition
    if user_input == "exit":
        print("Chatbot: Goodbye! All the best in your studies.")
        break

    # Greetings
    elif user_input in ["hi", "hello", "hey"]:
        print("Chatbot: Hello! How can I assist you today?")

    # CAT schedule
    elif "cat" in user_input and "schedule" in user_input:
        print("Chatbot: CATs are usually scheduled as per the academic calendar. Please check the notice board or student portal.")

    # Unit registration
    elif "unit" in user_input and "register" in user_input:
        print("Chatbot: Unit registration is done online via the student portal during the registration period.")

    # Default response
    else:
        print("Chatbot: Sorry, I didn't understand that. Please ask about CATs, units, or say hello.")
```
